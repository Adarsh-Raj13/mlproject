# ⚡ SMART LOAD MONITOR: Exploratory Data Analysis (EDA)
### Electrical Telemetry Signatures, Load Profiling & Fault Mode Exploration

This notebook explores the 12,000-sample electrical dataset simulated for 6 monitored electrical appliances under 7 distinct operating conditions.

```python
import os
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.figsize"] = (10, 5)
print("Libraries loaded successfully.")
```

## 1. Load Telemetry Datasets

```python
df_raw = pd.read_csv("../dataset/raw_data.csv")
df_proc = pd.read_csv("../dataset/processed_data.csv")

print(f"Raw dataset shape: {df_raw.shape}")
print(f"Processed dataset shape: {df_proc.shape}")
df_raw.head()
```

## 2. Statistical Distribution of Electrical Parameters

```python
features = ["Voltage", "Current", "Active_Power", "Power_Factor", "Temperature"]
df_raw[features].describe().round(3)
```

## 3. Appliance Class Distribution & Operational Conditions

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(data=df_raw, y="Appliance_Type", ax=axes[0], palette="viridis")
axes[0].set_title("Distribution of Telemetry Records by Appliance")

sns.countplot(data=df_raw, y="Operating_Condition", ax=axes[1], palette="magma")
axes[1].set_title("Distribution of Operating Conditions & Faults")

plt.tight_layout()
plt.show()
```

## 4. AC Power Relationships: Real (P) vs. Reactive (Q) on P-Q Plane

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_raw,
    x="Active_Power",
    y="Reactive_Power",
    hue="Appliance_Type",
    style="Operating_Condition",
    alpha=0.7,
    s=45
)
plt.title("Complex Power Plane (P vs Q) Across Monitored Loads")
plt.xlabel("Active Real Power P (Watts)")
plt.ylabel("Reactive Quadrature Power Q (VAR)")
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

## 5. Thermal Gradient vs. Current Draw Across Operating Conditions

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_raw,
    x="Current",
    y="Temperature",
    hue="Operating_Condition",
    palette="tab10",
    alpha=0.75,
    s=50
)
plt.title("Housing Temperature (°C) vs. Drawn Current (A)")
plt.xlabel("Current (Amperes RMS)")
plt.ylabel("Surface Temperature (°C)")
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

## 6. Feature Correlation Matrix

```python
numeric_cols = ["Voltage", "Current", "Active_Power", "Reactive_Power", "Apparent_Power", "Power_Factor", "Temperature", "Energy_Consumption"]
corr = df_raw[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", cbar=True, square=True)
plt.title("Correlation Matrix of Measured Telemetry Features")
plt.tight_layout()
plt.show()
```
